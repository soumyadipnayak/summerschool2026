```python
# Print even numbers from 2 to 20 using while loop
number = 2
while number <= 20:
    if number % 2 == 0:
        print(number, end="; ")
    number += 1
```

```python
# Calculate factorial of a number using while loop
# factorial of 5 = 5 * 4 * 3 * 2 * 1 = 120
n = int(input("Enter a number: "))
result = 1
while n >= 1:
    result *= n
    n -= 1
print(f"Factorial = {result}")
```

### 📝 Exercise 4.1: Loop Control Practice

```python
# Find first 5 prime numbers
# Use break when you have found 5 primes
from math import isqrt
count = 0
number = 2
while count < 5:
    if number == 2:
        count += 1
        print(number, end="; ")
    else:
        is_prime = True
        for divisor in range(2,isqrt(number)+1):
            if number % divisor == 0:
                is_prime = False
        if is_prime:
            count += 1
            print(number, end="; ")
    number += 1
            
```

### 🏆 Challenge 1: Armstrong Number Checker
Armstrong number: sum of cubes of digits equals the number  
Example: 153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153

```python
# Check if number is Armstrong number
num = int(input("Enter a number: "))
num_copy = num
num_of_digits = len(str(num))
sum_digits = 0
while num > 0:
    sum_digits += ((num%10)**3)
    num //= 10
if sum_digits == num_copy:
    print(f"{num_copy} is Armstrong!")
else:
    print(f"{num_copy} is NOT Armstrong!")
```

### 🏆 Challenge 2: GCD Calculator
Find Greatest Common Divisor using Euclidean algorithm

```python
# Calculate GCD of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
if a < b:
    a,b=b,a
while b > 0:
    a,b=b,a%b
print(f"GCD: {a}")
```

### 🏆 Challenge 3: Perfect Number Finder
Perfect number: sum of proper divisors equals the number  
Example: 6 = 1 + 2 + 3 (divisors of 6)

```python
from math import isqrt
num = int(input("Enter a number: "))
total = 1 if num > 1 else 0
for i in range(2, isqrt(num)+1):
    if num % i == 0:
        total += i
        if i != num//i:
            total += num//i
if total == num:
    print(f"{num} is perfect")
else:
    print(f"{num} is NOT perfect")
```

### 🏆 Challenge 4: Star Pattern - Hourglass

```python
# Create hourglass pattern
# *********
#  *******
#   *****
#    ***
#     *
#    ***
#   *****
#  *******
# *********
lines = int(input("Enter no. of lines=>"))
if lines % 2 != 0:
    for i in range(0, lines//2+1):
        print(" "*(i)+"* "*(int(lines/2 - i + 1)))
    for i in range(0, lines//2):
        print(" "*int(lines/2 - i - 1)+"* "*(i+2))  
else:
    for i in range(0, lines//2):
        print(" "*(i+1)+"* "*(int(lines/2 - i)))
    for i in range(0, lines//2):
        print(" "*int(lines/2 - i)+"* "*(i+1))
```

### 🏆 Challenge 5: Simple ATM System
Create a menu-driven ATM with:
- Check balance
- Deposit
- Withdraw
- Exit
Use a while loop for the menu

```python
# ATM System
balance = 10000
while True:
    choice = int(input("Enter your choice:\n1 for Balance Enquiry\n2 for Deposit \
    3 for withdrawal\n 0 to exit"))
    match choice:
        case 1:
            print(f"Your current balance is {balance:.2f}")
        case 2:
            amount = float(input("Confirm amount to be deposited=>"))
            if amount >= 0:
                balance += amount
                print(f"Deposit successful! Updated balance: {balance:.2f}")
            else:
                print("Please try again with a valid amount!")
        case 3:
            amount = float(input("Confirm amount to be withdrawn=>"))
            if amount >= 0 and balance >= amount:
                balance -= amount
                print(f"Withdrawal successful! Updated balance: {balance:.2f}")
            elif amount >= 0 and balance < amount:
                print("Insufficient balance, please try again")
            else:
                print("Please try again with a valid amount!")
        case 0:
            break
        case _:
            print("Invalid choice! Please try again!")
```

## 9. Week 1 Integration Project

### Final Project: Student Grade Management System

Create a complete program that:
1. Takes input for multiple students
2. Calculates grades based on marks
3. Displays statistics (average, highest, lowest)
4. Uses all concepts from Week 1

```python
# Complete Student Grade Management System
print("=== Student Grade Management System ===")

num_students = int(input("How many students? "))

# Initialize variables
total_marks = 0
highest = 0
lowest = 100
pass_count = 0

# Your complete solution here
# Use: variables, input, operators, conditionals, loops
for i in range(num_students):
    marks = int(input(f"Enter marks of student {i+1}"))
    if marks <= 30:
        print("Grade: F")
    elif marks <= 70:
        print("Grade: B")
        pass_count += 1
    else:
        print("Grade: A")
        pass_count += 1
    highest = marks if marks >= highest else highest
    lowest = marks if marks <= lowest else lowest
    total_marks += marks

print(f"Highest marks: {highest}\nLowest marks: {lowest}\nAvarage marks: {total_marks/num_students:.2f}, Total Passed Students: {pass_count}")
```
