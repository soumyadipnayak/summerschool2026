### Write a function `simple_interest(p, r, t)` that returns SI.

```python
def is_prime(number):
    if number <= 0:
        return False
    elif number == 1 or number == 2:
        return True
    else:
        for i in range(2, int(number**(1/2))+1):
            if number % i == 0:
                return False
        return True


number = int(input("Enter number=>"))
if is_prime(number):
    print(f"{number} is prime")
else:
    print(f"{number} is not prime")
```

### Write a function `stats(lst)` that returns (min, max, average).

```python
import random

nums = [random.randint(1, 100) for _ in range(10)]

def stats(nums):
    return (max(nums),min(nums),(sum(nums)/len(nums)))

max,min,avg = stats(nums)

print(nums)
print(f"Max: {max}, Min: {min}, Avarage: {avg}")
```
