## 1. Student Grade Management System

```python
students = {}

def add_student(students):
    name = input("Enter student name=>")
    students[name] = 0

def update_marks(students):
    name = input("Enter student name=>")
    marks = int(input("Enter marks obtained=>"))
    students[name] = marks

def display_report_card(students):
    name = input("Enter stdent name:")
    try:
        print(f"Student Name: {name}\nMarks Obtained: {students[name]}")
    except KeyError as e:
        print(f"Error occured! {e}\nPlease verify if the student exists!")

def calculate_avarage(students):
    avarage = sum(students.values())/len(students)
    return avarage

def display_topper(students):
    topper = max(students, key=students.get)
    print(f"Topper Name: {topper}\nMarks Obtained: {students[topper]}")

while True:
    user_choice = int(input("****Student Grade Management System****\n1 to add new student\n2 to add student marks\n3 to display report card\n4 to calculate avarage\n5 to display topper\n0 to exit\n=>"))
    match user_choice:
        case 1:
            add_student(students)
        case 2:
            update_marks(students)
        case 3:
            display_report_card(students)
        case 4:
            avarage = calculate_avarage(students)
            print(f"Avarage marks: {avarage}")
        case 5:
            display_topper(students)
        case 0:
            break
        case _:
            print("Please enter a valid choice!")
```
