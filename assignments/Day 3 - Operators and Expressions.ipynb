### 🏆 Challenge 1: Leap Year Checker
A year is leap if:
- Divisible by 4 AND not divisible by 100
- OR divisible by 400

```python
year = int(input("Enter year: "))
if (year%400 == 0) or (year%4 == 0 and year%100 != 0):
    print(f"{year} is leap year!")
else:
    print(f"{year} is NOT leap year!")
```

### 🏆 Challenge 2: Triangle Validator
Check if three sides can form a valid triangle:
- Sum of any two sides > third side
- All sides must be positive

```python
side1 = float(input("Enter side1 length: "))
side2 = float(input("Enter side2 length: "))
side3 = float(input("Enter side3 length: "))

if (side1 <= 0 or side2 <= 0 or side3 <= 0) or (side1+side2 <= side3) or (side1+side3 <= side2) or (side2+side3 <= side1):
    print(f"Invalid triangle with sides {side1}, {side2}, {side3}")
else:
    print(f"Valid triangle with sides {side1}, {side2}, {side3}")
```

### 🏆 Challenge 3: Complex Eligibility System
Job application eligibility:
- Age: 21-35
- Degree OR 5+ years experience
- Salary expectation < 100000
- Available to relocate OR remote work available

```python
age = int(input("Enter your age=>"))
degree = bool(input("Do you have an undergraduate degree from a recognised board/university?(1 for YES, 0 for NO)=>"))
experience = int(input("Enter total years of experience=>"))
salary_expectation = int(input("Enter your expected salary=>"))
relocation = bool(input(
    "Do you want to relocate?(1 for YES, 0 for NO)=>"))
remote_work = bool(input("Do you want to opt for remote work?(1 for YES, 0 for NO)=>"))
if (age >= 21 and age <= 35) and (degree or experience >= 5) and (salary_expectation < 100000) and (relocation or remote_work):
    print("Congratulations! You're eligible to apply!")
else:
    print("Unfortunately, you didn't qualify on one or more requirements!")
```

### 🏆 Challenge 4: Email Validator (Basic)
Check if email is valid:
- Contains @
- Contains .
- @ comes before .
- Length > 5

```python
email = input("Enter email: ")
if (len(email) > 5) and ("@" in email) and ("." in email) and (email.index("@") < email.index(".")) and (" " not in email):
    print("Email is valid")
else:
    print("Email is invalid")
```
