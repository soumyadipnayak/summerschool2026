## 1. Employee Performance Analyzer

```python
employees = {
    "Aman": 78,
    "Riya": 92,
    "Rahul": 55,
    "Priya": 81,
    "Karan": 48
}

def employee_analysis(employees):
    highest_score = list(employees.values())[0]
    highest_performer = ""
    total_score = 0
    count_under_60 = 0

    for name, score in employees.items():
        highest_score, highest_performer = (score, name) if score >= highest_score else (highest_score, highest_performer)
        total_score += score
        count_under_60 += 1 if score < 60 else 0
    avarage_score = total_score / len(employees.keys())
    above_avarage_employees = [k for k,v in employees.items() if v >= avarage_score]

    return highest_performer, avarage_score, above_avarage_employees, count_under_60


highest_performer, avarage_score, above_avarage_employees, count_under_60 = employee_analysis(employees)
print(f"Highest Performer: {highest_performer}\nAvarage Score: {avarage_score}\nEmployess above avarage: {", ".join(above_avarage_employees)}\nNo. of employess below 60 score: {count_under_60}")
```

## 2. Online Store Sales Support

```python
products = {
    "Laptop": 120,
    "Mouse": 85,
    "Keyboard": 60,
    "Monitor": 150,
    "Headphones": 95
}


def sales_report(products):
    best_selling_quantity = list(products.values())[0]
    best_selling_product = ""
    least_selling_quantity = list(products.values())[0]
    least_selling_product = ""
    total_sales = 0
    sales_greater_than_100 = []

    for product, units_sold in products.items():
        best_selling_product, best_selling_quantity = (product, units_sold) if units_sold >= best_selling_quantity else (best_selling_product, best_selling_quantity)
        least_selling_product, least_selling_quantity = (
            product, units_sold) if units_sold <= least_selling_quantity else (least_selling_product, least_selling_quantity)
        total_sales += units_sold
        if units_sold >= 100:
            sales_greater_than_100.append(product)
    
    return best_selling_product, least_selling_product, total_sales, sales_greater_than_100


best_selling_product, least_selling_product, total_sales, sales_greater_than_100 = sales_report(products)
print(f"Best selling product: {best_selling_product}\nLeast selling product: {least_selling_product}\nTotal units sold: {total_sales}\nProducts with more than 100 sales: {", ".join(sales_greater_than_100)}")
```

## 3. Student attendence management system

```python
all_students = ["Aman", "Riya", "Rahul", "Ankit", "Priya", "Karan"]
present_students = ["Aman", "Rahul", "Ankit", "Priya"]

def attendance_summary(all_students, present_students):
    absent_students = [student for student in all_students if student not in present_students]
    attendance_percentage = (len(present_students)/len(all_students))*100
    student_name_initial_a = [student for student in all_students if student[0]=="A"]
    attendence_greater_than_75 = True if attendance_percentage >= 75 else False
    return (absent_students,attendance_percentage,student_name_initial_a,attendence_greater_than_75)

absent_students, attendance_percentage, student_name_initial_a,attendence_greater_than_75 = attendance_summary(all_students, present_students)
print(f"Absent Students: {", ".join(absent_students)}\nAttendence: {attendance_percentage:.2f}%\nStudents with initial A: {", ".join(student_name_initial_a)}\nWhether attendence above 75: {attendence_greater_than_75}")
```

## 4. Library analytics system

```python
books = [
    "Python",
    "Java",
    "Python",
    "C++",
    "Java",
    "Machine Learning"
]

def library_report(books):
    total_books = len(books)
    unique_books = len(set(books))
    duplicate_books = books
    sorted_books = sorted(books)
    for book in set(books):
        duplicate_books.remove(book)
    return (total_books, unique_books, duplicate_books, sorted_books)


total_books, unique_books, duplicate_books, sorted_books = library_report(
    books)
print(f"Total no. of books: {total_books}\nTotal no. of unique books: {unique_books}\nDuplicate books: {duplicate_books}\nBooklist in alphabetical order: {sorted_books}")
```

## 5. University Result Processing System

```python
students = [
{"name": "Aman", "marks": 85},
{"name": "Riya", "marks": 95},
{"name": "Rahul", "marks": 38},
{"name": "Priya", "marks": 72},
{"name": "Karan", "marks": 58}
]

def results_analysis(students):
    maximum_marks = 0
    topper = ""
    total_marks = 0
    pass_count = 0

    for student in students:
        maximum_marks, topper = (int(student.get("marks",0)), student.get("name","")) if int(student.get("marks",0)) >= maximum_marks else (maximum_marks, topper)
        marks_obtained = int(student.get("marks",0))
        total_marks += marks_obtained
        if marks_obtained >= 40:
            pass_count += 1
        if marks_obtained >= 90:
            student["grade"] = "A"
        elif marks_obtained >= 75:
            student["grade"] ="B"
        elif marks_obtained >= 60:
            student["grade"] = "C"
        else:
            student["grade"] = "D"
    
    avarage_marks = total_marks/len(students)
    pass_percentage = (pass_count/len(students))*100
    above_avarage_students = [student.get(
        "name", "") for student in students if student.get("marks", 0) >= avarage_marks]
    
    return (topper,avarage_marks,above_avarage_students,pass_percentage,students)

topper, avarage_marks, above_avarage_students,pass_percentage,students = results_analysis(students)

print(f"Topper: {topper}\nAvarage Marks: {avarage_marks}\nStudents Above Avarage: {", ".join(above_avarage_students)}\nStudent Passed: {pass_percentage:.2f}%")
print("\nStudent Grades:")
for student in students:
    print(f"{student.get("name", "")}:{student.get("grade", "")}")
```
