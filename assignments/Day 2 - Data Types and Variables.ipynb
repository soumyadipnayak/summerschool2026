### 🏆 Challenge 1: Time Converter
Convert seconds to hours, minutes, and seconds

```python
# Example: 3665 seconds = 1 hour, 1 minute, 5 seconds
total_seconds = int(input("Enter total seconds: "))
hours = total_seconds // 3600
partial_hours = total_seconds % 3600
minutes = partial_hours // 60
seconds = partial_hours % 60
print(f"Time: {hours} hours, {minutes} minutes, {seconds} seconds")
```

### 🏆 Challenge 2: Area and Perimeter Calculator
Calculate for rectangle, circle, and triangle

```python
from math import pi
# Rectangle
print("Enter sides of rectangle")
x = float(input("x=>"))
y = float(input("y=>"))
print(f"Area of rectangle with sides {x} and {y} = {x*y:.5f}")
# Circle (use pi = 3.14159)
r = float(input("Enter radius of circle=>"))
print(f"Area of circle with radius {r} = {pi*(r**2):.5f}")
# Triangle
print("Enter sides of triangle")
a = float(input("a=>"))
b = float(input("b=>"))
c = float(input("c=>"))
s = (a+b+c)/2 # Half paremeter
print(
    f"Area of triangle with sides {a}, {b} and {c} = {(s*(s-a)*(s-b)*(s-c))**0.5:.5f}")
```

### 🏆 Challenge 4: Electricity Bill Calculator
Different rates for different unit ranges:
- First 100 units: ₹5 per unit
- Next 100 units: ₹7 per unit  
- Above 200: ₹10 per unit

(Use simple calculation for now, we'll improve with conditionals later)

```python
# Considering total unit consumed is greater thsn 200
total_units = int(input("Enter total units consumed=>"))

total_bill = ((total_units-200)*10)+(100*7)+(100*5)

print(f"Total bill for {total_units} units = ₹{total_bill}")
```
